## Question 2: Qualitative Data Visualization

### Dataset: Favorite Programming Language

| Language | Number of Students |
|---|---:|
| Python | 20 |
| Java | 10 |
| C++ | 8 |
| JavaScript | 12 |
| C# | 5 |

### Tasks
1. Identify the data type  
2. Create a **Bar Chart**  
3. Create a **Pie Chart**  
4. Write a short interpretation (2–3 sentences)


1. Dataset of favorite programming language is qualitative data. Language is also
qualitative. However, Number of students is quantitative.

```python
import matplotlib.pyplot as plt

languages = ["Python", "Java", "C++", "JavaScript", "C#"]
students = [20, 10, 8, 12, 5]

# Your code here

# Create a new figure (a blank canvas) that is 8 inches wide and 5 inches tall
plt.figure(figsize=(8, 5))

# Draw the bars: languages go on the x-axis, student counts set the bar heights
plt.bar(languages, students, color="steelblue")

# Add a title and labels so the chart is easy to understand
plt.title("Favorite Programming Language")
plt.xlabel("Programming Language")
plt.ylabel("Number of Students")

# Show the chart
plt.show()
```

```python
# Create a new figure that is 6 inches wide and 6 inches tall (a pie looks best as a square)
plt.figure(figsize=(6, 6))

# Draw the pie chart:
# students sets the size of each slice
# labels puts the language name next to each slice
# autopct shows each slice's percentage, rounded to 1 decimal place
# startangle=90 starts the first slice at the top
plt.pie(students, labels=languages, autopct="%1.1f%%", startangle=90)

# Add a title
plt.title("Favorite Programming Language")

# Show the chart
plt.show()
```

4. The data set shows the number of students in each of the 5 different programming
languages. Python is the most popular with 20 students choosing it and C# is the least
popular since there are only 5 students choose it. Overall, python and JavaScript are more
popular than any other programming languages.

## Question 3: Quantitative Data Visualization

### Dataset: Quiz Scores

65, 70, 72, 75, 78, 80, 81, 82, 84, 85, 86, 88, 90, 92, 95

### Tasks
1. Identify the data type  
2. Create a **Histogram**   
3. Describe the distribution


1. The data type of dataset quiz scores is quantitative because it stores numerical data.

```python
import matplotlib.pyplot as plt

scores = [65, 70, 72, 75, 78, 80, 81, 82, 84, 85, 86, 88, 90, 92, 95]

# Your code here

# Create a new figure that is 8 inches wide and 5 inches tall
plt.figure(figsize=(8, 5))

# Draw the histogram:
# scores is the data to count
# bins sets the edges of each group: 60-70, 70-80, 80-90, 90-100
# edgecolor draws a border around each bar so the bars are easy to tell apart
plt.hist(scores, bins=[60, 70, 80, 90, 100], color="steelblue", edgecolor="black")

# Add a title and labels
plt.title("Distribution of Quiz Scores")
plt.xlabel("Quiz Score")
plt.ylabel("Number of Students")

# Show the chart
plt.show()
```

3. The histogram shows that most students performed well. The tallest bar is the 80-90 group, with 7 of the 15 students, forming a clear peak in the distribution. The next group, 70-80, has 4 students, while only 1 student scored between 60 and 70 and 3 students scored between 90 and 100. This gives the distribution a hill-like, almost symmetric shape, with a slight tilt toward the lower scores, since 5 students scored below 80 and 3 scored 90 or above. Overall, there are no extreme or unusual scores, and all the values stay fairly close together, ranging from 65 to 95.

## Question 4: Relationship Between Variables

### Dataset: Study Hours vs Exam Score

| Student | Study Hours | Exam Score |
|---|---:|---:|
| A | 1 | 50 |
| B | 2 | 60 |
| C | 3 | 68 |
| D | 4 | 75 |
| E | 5 | 82 |
| F | 6 | 90 |

### Tasks
1. Create a **Scatter Plot**  
2. Explain the relationship between the variables


```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
exam_scores = [50, 60, 68, 75, 82, 90]

# Your code here

# Create a new figure that is 8 inches wide and 5 inches tall
plt.figure(figsize=(8, 5))

# Draw the scatter plot:
# study_hours goes on the x-axis, exam_scores goes on the y-axis
# each dot is one student
plt.scatter(study_hours, exam_scores, color="steelblue", s=80)

# Add a title and labels
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

# Show light grid lines so the dots are easier to read
plt.grid(True)

# Show the chart
plt.show()
```

2. It is a strong positive relationship between study hours and exam scores because study
hours increase, exam scores consistently increase as well, and the points fall closely
along a straight upward sloping line.
