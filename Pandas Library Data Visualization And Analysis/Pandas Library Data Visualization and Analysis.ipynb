<details>
<summary>Click to see more about <span style="color:orange;">df.duplicated()</span></summary>

The `df.duplicated()` function checks each row in the DataFrame and returns `True` if the row is a duplicate of a previous row, and `False` otherwise.  
When combined with `.sum()`, it counts the total number of duplicate rows in the dataset.  

This is useful to identify duplicate entries that may need to be **removed** to ensure accurate analysis.

For more details and examples, visit the official Pandas documentation:  
[Learn more about `df.duplicated()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.duplicated.html)

##### 6.1 Histogram: Distribution of salary_usd

````python
x = df["salary_usd"]
plt.figure(figsize=(8, 5))
sns.histplot(x, bins=20, kde=True)
plt.title("Distribution of Salaries (USD)")
plt.xlabel("Salary (USD)")
plt.ylabel("Number of Jobs")
plt.show()
````

##### 6.2 Barplot: Compare number posts for each jobs


````python
# Number of posts for each jobs

'''
df["job_title"].value_counts()
df["job_title"].value_counts().plot(kind="bar")'''


# Count of each job title
job_counts = df["job_title"].value_counts()

# x = job titles, y = counts
x = job_counts.index # job-titles
y = job_counts.values # job-count

# Plot bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x=y, y=x)  # horizontal bar chart
#sns.barplot(x=x, y=y)
plt.title("Number of Job Postings by Job Title")
plt.xlabel("Number of Jobs")
plt.ylabel("Job Title")
plt.show()
````

##### 6.3 Bar Chat: Compare which job has the highest average salary



````python

# 1. Group by job title and calculate average salary
avg_salary_by_job = df.groupby("job_title")["salary_usd"].mean()

# 2. Sort from highest to lowest
avg_salary_by_job_sorted = avg_salary_by_job.sort_values(ascending=False)

# 3. Display the top 5 highest-paying job titles
print(avg_salary_by_job_sorted.head(5))

# Plot graph
# Bar chart of top 5 highest-paying job titles
plt.figure(figsize=(10, 6))
sns.barplot(x=avg_salary_by_job_sorted.head(5).values,
            y=avg_salary_by_job_sorted.head(5).index,
            palette="viridis")
plt.title("Top 5 Highest-Paying Job Titles")
plt.xlabel("Average Salary (USD)")
plt.ylabel("Job Title")
plt.show()
````

##### 6.4 Distribution of salary currencies

````python
# Counts how many jobs are listed in each currency.

# Count number of jobs per currency
currency_counts = df["salary_currency"].value_counts()

# x = categories (currencies), y = counts
x = currency_counts.index      # ['USD', 'EUR', 'GBP', ...]
y = currency_counts.values     # [300, 120, 50, ...]

# Create bar chart
plt.figure(figsize=(6,4))
plt.bar(x, y, color="blue")   # x = categories, y = counts
plt.title("Number of Job Postings by Salary Currency")
plt.xlabel("Currency")
plt.ylabel("Number of Jobs")
plt.show()
````

````python
#1. Plot a graph to compare which job titles require higher Python programming skill.

# Check each job posting: does its required_skills text contain the word "Python"?
# This gives True (it does) or False (it does not) for every row.
has_python = df["required_skills"].str.contains("Python")

# Keep only the postings that require Python,
# then count how many postings each job title has
python_counts = df[has_python]["job_title"].value_counts()

# Print the numbers so we can read them
print(python_counts)

# Draw a horizontal bar chart (long job titles are easier to read this way)
plt.figure(figsize=(10, 6))
sns.barplot(x=python_counts.values, y=python_counts.index)

# Add a title and labels
plt.title("Number of Job Postings Requiring Python, by Job Title")
plt.xlabel("Number of Jobs Requiring Python")
plt.ylabel("Job Title")

# Show the chart
plt.show()
````

````python
#2. plot a graph to compare which job title require python and R programming skills together.

# Split each skills text into a list of separate skills
# Example: "Scala, SQL, Python" becomes ["Scala", "SQL", "Python"]
skills_list = df["required_skills"].str.split(", ")

# True if the exact skill "Python" is in the list, otherwise False
has_python = skills_list.apply(lambda skills: "Python" in skills)

# True if the exact skill "R" is in the list, otherwise False
has_r = skills_list.apply(lambda skills: "R" in skills)

# Keep only the postings that require BOTH Python and R (the & means "and")
both = df[has_python & has_r]

# Count how many of these postings each job title has
both_counts = both["job_title"].value_counts()

# Print the numbers so we can read them
print("Postings requiring both Python and R:", len(both))
print(both_counts)

# Draw a horizontal bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x=both_counts.values, y=both_counts.index)

# Add a title and labels
plt.title("Number of Job Postings Requiring Both Python and R, by Job Title")
plt.xlabel("Number of Jobs Requiring Python and R")
plt.ylabel("Job Title")

# Show the chart
plt.show()
````

````python
#3. How much maximum salary for Machine Learning Engineer or Machine Learning Researcher for entry level in Switzerland?

# Condition 1: the job title is one of the two titles (isin checks "is it in this list?")
is_ml_job = df["job_title"].isin(["Machine Learning Engineer", "Machine Learning Researcher"])

# Condition 2: the experience level is entry level (EN)
is_entry = df["experience_level"] == "EN"

# Condition 3: the company is located in Switzerland
is_swiss = df["company_location"] == "Switzerland"

# Keep only the rows where ALL three conditions are True (the & means "and")
matches = df[is_ml_job & is_entry & is_swiss]

# Show how many matching jobs we found
print("Number of matching jobs:", len(matches))

# Show the matching jobs, highest salary first
print(matches[["job_title", "experience_level", "company_location", "salary_usd"]]
      .sort_values("salary_usd", ascending=False))

# The answer: the highest salary among the matching jobs
print()
print("Maximum salary (USD):", matches["salary_usd"].max())
````

````python
#4. Plot graph to see the relationship between salary and years experience
# (column name in dataset: years_experience) for all type of AI job(AI Research Scientist, AI Software Engineer,etc).

# Create a new figure that is 10 inches wide and 6 inches tall
plt.figure(figsize=(10, 6))

# Draw the scatter plot:
# x-axis = years of experience, y-axis = salary
# alpha=0.3 makes each dot see-through, so crowded areas look darker
# s=15 makes the dots small
sns.scatterplot(data=df, x="years_experience", y="salary_usd", alpha=0.3, s=15)

# Add a title and labels
plt.title("Salary vs Years of Experience (All AI Jobs)")
plt.xlabel("Years of Experience")
plt.ylabel("Salary (USD)")

# Show light grid lines so the dots are easier to read
plt.grid(True)

# Show the chart
plt.show()
````

````python
#5. Compare average salary of Machine Learning Engineer job with different required education(columns name: education_required)

# Keep only the Machine Learning Engineer jobs
ml_engineers = df[df["job_title"] == "Machine Learning Engineer"]

# Group these jobs by the education they require,
# then calculate the average salary of each group
avg_salary_by_education = ml_engineers.groupby("education_required")["salary_usd"].mean()

# Sort from highest to lowest so the chart is easy to read
avg_salary_by_education = avg_salary_by_education.sort_values(ascending=False)

# Print the numbers so we can read them
print(avg_salary_by_education)

# Draw a bar chart
plt.figure(figsize=(8, 5))
sns.barplot(x=avg_salary_by_education.index, y=avg_salary_by_education.values)

# Add a title and labels
plt.title("Average Salary of Machine Learning Engineers by Required Education")
plt.xlabel("Required Education")
plt.ylabel("Average Salary (USD)")

# Show the chart
plt.show()
````
