```python
import numpy as np
```

```python
#1. Train data
data = np.array([
    [21.8, 3.4, 279, 0],
    [29.9, 5.8, 240, 0],
    [34.4, 8.1, 285, 0],
    [30.3, 8.9, 251, 1],
    [36.9, 8.6, 197, 0],
    [31.1, 4.1, 290, 0],
    [21.0, 3.4, 268, 0],
    [22.1, 8.4, 256, 0],
    [25.9, 4.6, 287, 1],
    [38.2, 6.5, 230, 1],
    [28.1, 9.4, 208, 1],
    [37.2, 4.5, 289, 1],
    [19.6, 3.8, 299, 1],
    [33.1, 4.6, 262, 1],
    [28.3, 4.2, 161, 0],
    [21.8, 5.8, 189, 1],
    [19.8, 5.7, 259, 0],
    [24.9, 4.6, 264, 1],
    [36.1, 9.5, 259, 0],
    [32.1, 4.3, 288, 0]
])
X_train = data[:, :3]
y_train = data[:, 3].astype(int)
n = len(X_train)
```

```python
#2. Input parameter
while True:
    k = int(input(f"Enter k (0 < k <= {n}): "))
    if 0 < k <= n:
        break
    print("Invalid k, please try again.")
```

```python
#3. Distance calculation
def calculate_distance(point1, point2):
    distance = 0.0
    for i in range(len(point1)):
        distance += (point1[i] - point2[i]) ** 2
    return distance ** 0.5
```

```python
#4. Test sample
test = np.array([20.0, 3.4, 161.0])
```

```python
#5. Iterate over the training samples to calculate distance
distances = np.zeros(n)
for i in range(n):
    distances[i] = calculate_distance(X_train[i], test)

print("No.  F1     F2    F3    Label  Distance")
for i in range(n):
    print(f"{i+1:<4} {X_train[i,0]:<6} {X_train[i,1]:<5} {X_train[i,2]:<5.0f} {y_train[i]:<6} {distances[i]:.4f}")
```

```python
#7. Nearest neighbor
nearest_idx = np.argsort(distances)[:k]
k_smallest_distances = distances[nearest_idx]
k_smallest_labels = y_train[nearest_idx]
print("Neighbor samples (No.):", nearest_idx + 1)
print("Distances of KNN:", k_smallest_distances)
print("Labels of KNN:", k_smallest_labels)
```

```python
#8. Compute the label
epsilon = 1e-5
weights = 1 / (k_smallest_distances + epsilon)
print("Weights:", weights)

weighted_votes = {}
for c in np.unique(y_train):
    weighted_votes[c] = np.sum(weights[k_smallest_labels == c])
print("Weighted votes:", weighted_votes)

predicted_label = max(weighted_votes, key=weighted_votes.get)
print("Predicted label:", predicted_label)
```
